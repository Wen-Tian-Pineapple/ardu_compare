```python
import rosbag
import numpy as np

import os

# Define the path to your ROS bag file
bag_file_path = '/home/xxh/Doublebee/bb_ws/2024-05-27-13-41-17.bag'

# Open the ROS bag file
bag = rosbag.Bag(bag_file_path)

# List to store the data
data_list = []

# Read the messages from the specified topic
for topic, msg, t in bag.read_messages(topics=['/mavros/jai_raw']):
    # Extract the fields from the Jai_Raw message
    data = [
        msg.time_usec,
        msg.rol,
        msg.pit,
        msg.yaw,
        msg.drol,
        msg.dpit,
        msg.dyaw,
        msg.wr,
        msg.wl,
        msg.ur,
        msg.ul,
        msg.disired_pit,
        msg.desired_dpit,
        msg.beta,
        msg.Thr_out
    ]
    data_list.append(data)

# Convert the list to a NumPy array
data_array = np.array(data_list)

# Define the output directory and file name
output_dir = 'output'
os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, 'jai_raw_data.npy')



# Save the NumPy array to a file
np.save(output_file, data_array)

# Close the bag file
bag.close()

print(f"Data has been saved to {output_file} with shape {data_array.shape}")
```
